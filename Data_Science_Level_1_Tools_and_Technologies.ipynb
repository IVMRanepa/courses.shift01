### Задача 1. Посчитать описательные статистики.

```python
# Задача 1. Посчитать описательные статистики.
# Комментарий: используем классические статистические методы из EDA и проверки гипотез.

np.random.seed(201)
control = np.random.normal(loc=100, scale=12, size=60)
test = np.random.normal(loc=106, scale=12, size=60)

df = pd.DataFrame({"group": ["control"]*60 + ["test"]*60,
                   "value": np.concatenate([control, test])})

print(df.groupby("group")["value"].agg(["mean", "std", "median"]))

stat, pvalue = stats.ttest_ind(control, test, equal_var=False)
print("t-stat:", round(stat, 3))
print("p-value:", round(pvalue, 5))

sns.boxplot(data=df, x="group", y="value")
plt.title("Сравнение двух групп")
plt.show()

# Вывод:
# Если p-value < 0.05, различия между группами статистически значимы.
```

### Задача 1. Использовать PCA для снижения размерности.

```python
# Задача 1. Использовать PCA для снижения размерности.
# Комментарий: применяем дополнительные инструменты — PCA и кластеризацию.

np.random.seed(901)
X = pd.DataFrame(np.random.normal(size=(150, 5)), columns=["f1", "f2", "f3", "f4", "f5"])
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
components = pca.fit_transform(X_scaled)
km = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = km.fit_predict(components)

viz = pd.DataFrame({"pc1": components[:, 0], "pc2": components[:, 1], "cluster": labels})
sns.scatterplot(data=viz, x="pc1", y="pc2", hue="cluster", palette="deep")
plt.title("Кластеры в пространстве главных компонент")
plt.show()
print("Explained variance ratio:", pca.explained_variance_ratio_)

# Вывод:
# PCA помогает уменьшить размерность, а KMeans — выделить группы наблюдений.
```
